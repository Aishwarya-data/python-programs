## Level 1.print if the user guessed number is right or not.

```python
# variable

secret_number = 39

# Get input from user:

user_input = int(input("Enter your guess:") )

if user_input == secret_number:
    print("Congratulations! your guess is right.")
else:
    print("Sorry! your guess is wrong")
```

## Level 2.print if the user guessed number is right or not with maximum of 7 trials.

### 2.1 using for loop -- when the end number is defined

```python
# variable

secret_number = 39

# Get input from user:
for i in range(0,7):
    user_input = int(input("Enter your guess:") )
    if user_input == secret_number:
        print("Congratulations! your guess is right.")
        break 
    else:
        print("Sorry! your guess is wrong")
```

### 2.2 using while loop when the end number is not defined and the instructions has to execute infinite times

```python
# variable

secret_number = 39

# Get input from user:
while True:
    user_input = int(input("Enter your guess:") )
    if user_input == secret_number:
        print("Congratulations! your guess is right.")
        break 
    else:
        print("Sorry! your guess is wrong")
```

### DIFFERENCE BETWEEN FOR AND WHILE LOOP

* when the number of iteration is exactly defined for loop is used.
*  when the number of iteration is not defined or when infinite times needed while loop is used.
*  contrl statements like break,continue,pass is used with loops

## Level 3.Give Multiple conditions

1. user should give number between 1 to 100. If the given number is wrong the program should say "Invalid number".
2. If the user guessed number is greater than the secret number, the program should say "Your number is greater than the secret number"
3. If the user guessed number is lesser than the secret number, the program should say "Your number is lesser than the secret number"

```python
print("Welcome to Number Guessing Game!!!")
print("Guess number between 1 to 100")

secret_number = 24 

for i in range(1,8):       # 7 times loop will execute
    user_input = int(input("Enter Your Guessed number:"))
    if user_input <1 or user_input > 100:
        print("Invalid Number ! please enter number between 1 to 100 ")
    elif user_input > secret_number:
        print("Your number is greater than the secret number")
    elif user_input < secret_number:
        print("Your number is lesser than the secret number")
    else:
        print("Congratulations! your guess is right.")
        print(f"Well Done! You guessed the {secret_number} in {i} attempts .")
        break
```

## Level 7 : How to handle invalid characters or values

#### Exception handling to handle errors

```python
import random
print("Welcome to Number Guessing Game!!!")
print("Guess number between 1 to 100")

secret_number = random.randint(1,100)

for i in range(1,8):       # 7 times loop will execute
    try:
        user_input = int(input("Enter Your Guessed number:"))
        if user_input <1 or user_input > 100:
            print("Invalid Number ! please enter number between 1 to 100 ")
        elif user_input > secret_number:
            print("Your number is greater than the secret number")
        elif user_input < secret_number:
            print("Your number is lesser than the secret number")
        else:
            print("Congratulations! your guess is right.")
            break
    except ValueError:
        print("Invalid characters not recognized")
if secret_number == user_input:
    print(f"Well Done! You guessed the {secret_number} in {i} attempts .")
else:
    print("Sorry! All your trials exceeded, Try again !!!")
```

## Level 8: Take this program to run as a function when it is called

1. Define the function
2. call the function

```python
def number_guessing_game():
    import random
    print("Welcome to Number Guessing Game!!!")
    print("Guess number between 1 to 100")
    
    secret_number = random.randint(1,100)
    
    for i in range(1,8):       # 7 times loop will execute
        try:
            user_input = int(input("Enter Your Guessed number:"))
            if user_input <1 or user_input > 100:
                print("Invalid Number ! please enter number between 1 to 100 ")
            elif user_input > secret_number:
                print("Your number is greater than the secret number")
            elif user_input < secret_number:
                print("Your number is lesser than the secret number")
            else:
                print("Congratulations! your guess is right.")
                break
        except ValueError:
            print("Invalid characters not recognized")
    if secret_number == user_input:
        print(f"Well Done! You guessed the {secret_number} in {i} attempts .")
    else:
        print("Sorry! All your trials exceeded, Try again !!!")
```
