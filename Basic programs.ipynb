## 1.reverse a number

* Extract last digit using %
* Build reverse using *10 + digit
* Remove digit using //

```python
n = int(input("Enter number to reverse:"))

original_number = n   
reverse = 0

while n >0:
    digit = n%10                     # Extract last digit using %
    reverse = reverse*10 +digit      # Build reverse using *10 + digit
    n = n//10                        # Removes last digit
print(f"The reverse of {original_number} is",reverse)
```

## 2.Palindrome number

* Key Logic (Same as Reverse)

* only thing if else compare orginal number with reverse number [if original_number == reverse:]

```python
n = int(input("Enter number to check:"))

original_number = n   
reverse = 0

while n >0:
    digit = n%10                     # Extract last digit using %
    reverse = reverse*10 +digit      # Build reverse using *10 + digit
    n = n//10                        # Removes last digit

if original_number == reverse:
    print(f"{original_number} is Palindrome")
else:
    print(f"{original_number} is not Palindrome")
```

## 2.prime number

A number divide by 1 and itself

n % i == 0 → divisible → not prime

break → loop stop

else (loop) → no divisor → prime

```python
n = int(input("Enter a number to check prime: "))

if n<=1:
    print("Not a prime Number")
else:
    for i in range(2,n):
        if n%i == 0:      
            print("Not a prime Number")
            break
    else:
            print("prime Number")
            
```

## 4.Armstrong number

sum of power of digits = original number

Digits: 1, 5, 3  
Power = 3 (3 digits)

1³ + 5³ + 3³  
= 1 + 125 + 27  
= 153 

```python
n = int(input("Enter number to check:"))

original_number = n
power = len(str(n))
total =0
while n>0:
    digit = n%10
    total += digit**power
    n=n//10

if original_number == total:
    print("Armstrong Number")
else:
    print("Not Armstrong")
```

## 2.Factorial

```python
def factorial(n):
    fact = 1
    for i in range(1, n+1):
        fact *= i
    return fact

print("Factorial:", factorial(5))
```

## 3.Palindrome using Function

```python
def is_palindrome(n):
    original = n
    reverse = 0

    while n > 0:
        reverse = reverse * 10 + (n % 10)
        n //= 10

    return original == reverse

num = int(input("Enter number: "))

if is_palindrome(num):
    print("Palindrome ")
else:
    print("Not Palindrome ")
```
